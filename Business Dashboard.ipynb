```python
import pandas as pd

skills_df = pd.read_csv("jobs_with_skills.csv")

print("Skills data loaded successfully")
print(skills_df.head())
print(skills_df.columns)
```

```python
import ast
import pandas as pd

# Convert extracted_skills from text into Python lists
skills_df["extracted_skills"] = skills_df["extracted_skills"].apply(ast.literal_eval)

# Create one row for each skill
skill_data = skills_df.explode("extracted_skills")

# Rename the skill column
skill_data = skill_data.rename(columns={
    "extracted_skills": "skill"
})

# Keep the columns needed for the dashboard
skill_data = skill_data[
    ["job_id", "job_title", "company", "location", "experience", "education", "salary", "job_type", "skill"]
]

print("Skill table created successfully")
print(skill_data.head(10))
print("Total skill records:", len(skill_data))
```

```python
# Create skill category mapping

skill_category = {
    "python": "Programming",
    "java": "Programming",
    "c++": "Programming",
    "javascript": "Programming",
    
    "sql": "Database",
    "mysql": "Database",
    "postgresql": "Database",
    "mongodb": "Database",
    
    "power bi": "Data Analytics",
    "tableau": "Data Analytics",
    "excel": "Data Analytics",
    
    "pandas": "Data Science",
    "numpy": "Data Science",
    
    "scikit-learn": "Machine Learning",
    "tensorflow": "Machine Learning",
    "pytorch": "Machine Learning",
    
    "aws": "Cloud",
    "azure": "Cloud",
    "gcp": "Cloud",
    
    "spark": "Big Data",
    "hadoop": "Big Data",
    
    "docker": "DevOps",
    "kubernetes": "DevOps",
    
    "git": "Version Control"
}

# Add category column
skill_data["skill_category"] = skill_data["skill"].map(skill_category)

print("Skill categories added successfully")
print(skill_data[["skill", "skill_category"]].head(15))
```

```python
print("Skills without category:")
print(skill_data[skill_data["skill_category"].isna()]["skill"].unique())
```

```python
skill_data.to_csv("job_skill_data.csv", index=False)

print("job_skill_data.csv saved successfully")
```

```python
missing_skills = skill_data[
    skill_data["skill_category"].isna()
]["skill"].value_counts()

print("Missing skills:")
print(missing_skills)
```
