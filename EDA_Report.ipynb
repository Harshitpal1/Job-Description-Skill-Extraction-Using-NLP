```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("raw_jobs_10k.csv")
```

```python
df["description_length"] = df["job_description"].str.len()
```

```python
print("Total Jobs:", len(df))
```

```python
print("Unique Job Titles:", df["job_title"].nunique())
```

```python
print("Average Description Length:", df["description_length"].mean())
```

```python
print("Minimum Description Length:", df["description_length"].min())
```

```python
print("Maximum Description Length:", df["description_length"].max())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df["job_title"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")
plt.show()
```

```python
df["location"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Job Locations")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
plt.show()
```

```python
plt.hist(df["description_length"], bins=30)

plt.title("Job Description Length Distribution")
plt.xlabel("Description Length")
plt.ylabel("Number of Jobs")
plt.show()
```

```python
df.isnull().sum().plot(kind="bar")

plt.title("Missing Values")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.show()
```

```python
df["company"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Companies")
plt.xlabel("Company")
plt.ylabel("Number of Jobs")
plt.show()
```
