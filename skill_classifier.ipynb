```python
import pandas as pd

data = {
    "text": [
        "Python",
        "Python programming",
        "Java",
        "Java programming",
        "AWS",
        "Amazon Web Services",
        "Azure",
        "SQL",
        "MySQL",
        "PostgreSQL",
        "Power BI",
        "Tableau",
        "Excel",
        "TensorFlow",
        "PyTorch",
        "Docker",
        "Kubernetes"
    ],

    "label": [
        "SKILL",
        "SKILL",
        "SKILL",
        "SKILL",
        "CLOUD",
        "CLOUD",
        "CLOUD",
        "DATABASE",
        "DATABASE",
        "DATABASE",
        "BI_TOOL",
        "BI_TOOL",
        "BI_TOOL",
        "TECHNOLOGY",
        "TECHNOLOGY",
        "TECHNOLOGY",
        "TECHNOLOGY"
    ]
}

train_df = pd.DataFrame(data)

train_df
```

```python
X = train_df["text"]
y = train_df["label"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression())
])
```

```python
model.fit(X, y)

print("Model training completed!")
```

```python
test_data = [
    "Python programming",
    "AWS cloud",
    "PostgreSQL database",
    "Power BI dashboard",
    "Docker container"
]

predictions = model.predict(test_data)

for text, prediction in zip(test_data, predictions):
    print(text, "->", prediction)
```

```python
import joblib

joblib.dump(model, "skill_classifier.pkl")

print("Model saved successfully!")
```
