```python
import pandas as pd
from collections import Counter
```

```python
df = pd.read_csv("clean_jobs.csv")

print(df.head())
print(df.shape)
```

```python
skills = [
    "python",
    "java",
    "c++",
    "javascript",
    "sql",
    "mysql",
    "postgresql",
    "mongodb",
    "power bi",
    "tableau",
    "excel",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "pytorch",
    "aws",
    "azure",
    "gcp",
    "spark",
    "hadoop",
    "docker",
    "kubernetes",
    "git"
]
```

```python
def extract_skills(text):
    found = []

    text = text.lower()

    for skill in skills:
        if skill in text:
            found.append(skill)

    return found
```

```python
df["extracted_skills"] = df["clean_description"].apply(extract_skills)

print(df[["job_title", "extracted_skills"]].head(10))
```

```python
all_skills = []

for skill_list in df["extracted_skills"]:
    all_skills.extend(skill_list)

skill_counts = Counter(all_skills)

print(skill_counts.most_common(10))
```

```python
for skill, count in skill_counts.most_common():
    print(skill, ":", count)
```

```python
df.to_csv("jobs_with_skills.csv", index=False)

print("File saved successfully!")
```
