```python
import pandas as pd
import re

print("Libraries loaded successfully")
```

```python
normalization_map = {

    # Python
    "python": "Python",
    "python3": "Python",
    "python 3": "Python",
    "python programming": "Python",

    # Java
    "java": "Java",

    # C++
    "c++": "C++",
    "cpp": "C++",

    # JavaScript
    "javascript": "JavaScript",
    "js": "JavaScript",

    # TypeScript
    "typescript": "TypeScript",
    "ts": "TypeScript",

    # SQL
    "sql": "SQL",

    # MySQL
    "mysql": "MySQL",

    # PostgreSQL
    "postgres": "PostgreSQL",
    "postgresql": "PostgreSQL",

    # MongoDB
    "mongodb": "MongoDB",
    "mongo": "MongoDB",

    # Power BI
    "powerbi": "Power BI",
    "power bi": "Power BI",
    "power-bi": "Power BI",

    # Tableau
    "tableau": "Tableau",

    # Excel
    "excel": "Excel",
    "ms excel": "Excel",

    # Data Science
    "pandas": "Pandas",
    "numpy": "NumPy",

    # Machine Learning
    "scikit learn": "Scikit-learn",
    "scikit-learn": "Scikit-learn",
    "sklearn": "Scikit-learn",

    "tensorflow": "TensorFlow",
    "pytorch": "PyTorch",

    # ML Concepts
    "machine learning": "Machine Learning",
    "deep learning": "Deep Learning",
    "natural language processing": "Natural Language Processing",

    # Cloud
    "aws": "AWS",
    "azure": "Azure",
    "gcp": "GCP",
    "google cloud": "GCP",

    # Big Data
    "spark": "Spark",
    "apache spark": "Spark",

    "hadoop": "Hadoop",
    "apache hadoop": "Hadoop",

    # DevOps
    "docker": "Docker",
    "kubernetes": "Kubernetes",
    "k8s": "Kubernetes",

    # Version Control
    "git": "Git"
}

print("Normalization dictionary created")
print("Total aliases:", len(normalization_map))
```

```python
def normalize_skill(raw_skill):

    # Convert to string
    skill = str(raw_skill)

    # Lowercase
    skill = skill.lower().strip()

    # Remove extra spaces
    skill = re.sub(r"\s+", " ", skill)

    # Check alias dictionary
    if skill in normalization_map:
        return normalization_map[skill]

    # If skill is not found
    return skill.title()
```

```python
test_skills = [
    "Python",
    "Python3",
    "Python 3",
    "Python programming",
    "PowerBI",
    "Power BI",
    "Power-BI",
    "postgres",
    "postgresql",
    "sklearn",
    "k8s",
    "google cloud"
]

for skill in test_skills:

    normalized = normalize_skill(skill)

    print(
        skill,
        "->",
        normalized
    )
```

```python
normalized_skills_df = pd.DataFrame(
    list(normalization_map.items()),
    columns=[
        "raw_skill",
        "canonical_skill"
    ]
)

normalized_skills_df = normalized_skills_df.drop_duplicates()

normalized_skills_df = normalized_skills_df.sort_values(
    "canonical_skill"
).reset_index(drop=True)

print(normalized_skills_df.head(20))
```

```python
normalized_skills_df = pd.DataFrame(
    list(normalization_map.items()),
    columns=[
        "raw_skill",
        "canonical_skill"
    ]
)

normalized_skills_df = normalized_skills_df.drop_duplicates()

normalized_skills_df = normalized_skills_df.sort_values(
    "canonical_skill"
).reset_index(drop=True)

print(normalized_skills_df.head(20))
```

```python
normalized_skills_df.to_csv(
    "normalized_skills.csv",
    index=False
)

print("normalized_skills.csv saved successfully")
```
