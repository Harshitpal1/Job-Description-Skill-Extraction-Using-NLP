```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("clean_jobs.csv")

print(df.shape)
df.head()
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
text1 = "ML"
text2 = "Machine Learning"

embedding1 = model.encode([text1])
embedding2 = model.encode([text2])
```

```python
similarity = cosine_similarity(
    embedding1,
    embedding2
)

print("Similarity:", similarity[0][0])
```

```python
pairs = [
    ("ML", "Machine Learning"),
    ("NLP", "Natural Language Processing"),
    ("AI", "Artificial Intelligence"),
    ("Python developer", "Python programming"),
    ("database", "SQL")
]

for text1, text2 in pairs:

    emb1 = model.encode([text1])
    emb2 = model.encode([text2])

    score = cosine_similarity(
        emb1,
        emb2
    )[0][0]

    print(text1, "<->", text2)
    print("Similarity:", round(score, 4))
    print()
```

```python
skills = [
    "Python",
    "SQL",
    "Java",
    "Machine Learning",
    "Deep Learning",
    "Natural Language Processing",
    "Power BI",
    "Data Science",
    "AWS",
    "Docker",
    "Kubernetes",
    "PostgreSQL"
]
```

```python
skill_embeddings = model.encode(skills)

print("Skill embeddings created.")
```

```python
phrase = "ML"
```

```python
phrase_embedding = model.encode([phrase])
```

```python
similarities = cosine_similarity(
    phrase_embedding,
    skill_embeddings
)[0]
```

```python
results = pd.DataFrame({
    "skill": skills,
    "similarity": similarities
})

results = results.sort_values(
    by="similarity",
    ascending=False
)

print(results)
```

```python
best_skill = results.iloc[0]["skill"]
best_score = results.iloc[0]["similarity"]

print("Best Match:", best_skill)
print("Similarity:", round(best_score, 4))
```

```python
threshold = 0.35
```

```python
if best_score >= threshold:
    print("Possible Skill:", best_skill)
else:
    print("No strong skill match found.")
```

```python
results.to_csv(
    "semantic_similarity_results.csv",
    index=False
)
```
